```python
#dictionary is ai generated
soccer_players = {
    "Lionel Messi": {
        "team": "Inter Miami",
        "league": "MLS",
        "season": "2025",
        "matches": 28,
        "minutes": 2420,
        "goals": 29,
        "assists": 19
    },
    "Cristiano Ronaldo": {
        "team": "Al Nassr",
        "league": "Saudi Pro League",
        "season": "2025-26",
        "matches": 30,
        "minutes": 2610,
        "goals": 28,
        "assists": 2
    },
    "Kylian Mbappe": {
        "team": "Real Madrid",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 31,
        "minutes": 2599,
        "goals": 25,
        "assists": 5
    },
    "Erling Haaland": {
        "team": "Manchester City",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 35,
        "minutes": 2953,
        "goals": 27,
        "assists": 8
    },
    "Lamine Yamal": {
        "team": "Barcelona",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 28,
        "minutes": 2262,
        "goals": 16,
        "assists": 11
    },
    "Harry Kane": {
        "team": "Bayern Munich",
        "league": "Bundesliga",
        "season": "2025-26",
        "matches": 31,
        "minutes": 2377,
        "goals": 36,
        "assists": 5
    },
    "Vinicius Junior": {
        "team": "Real Madrid",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 36,
        "minutes": 2815,
        "goals": 16,
        "assists": 5
    },
    "Mohamed Salah": {
        "team": "Liverpool",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 27,
        "minutes": 2146,
        "goals": 7,
        "assists": 7
    },
    "Robert Lewandowski": {
        "team": "Barcelona",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 31,
        "minutes": 1664,
        "goals": 14,
        "assists": 2
    },
    "Raphinha": {
        "team": "Barcelona",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 22,
        "minutes": 1381,
        "goals": 13,
        "assists": 3
    },
    "Jude Bellingham": {
        "team": "Real Madrid",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 28,
        "minutes": 1916,
        "goals": 6,
        "assists": 4
    },
    "Pedri": {
        "team": "Barcelona",
        "league": "La Liga",
        "season": "2025-26",
        "matches": 29,
        "minutes": 2104,
        "goals": 2,
        "assists": 9
    },
    "Bukayo Saka": {
        "team": "Arsenal",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 31,
        "minutes": 2222,
        "goals": 7,
        "assists": 5
    },
    "Cole Palmer": {
        "team": "Chelsea",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 26,
        "minutes": 1956,
        "goals": 10,
        "assists": 1
    },
    "Florian Wirtz": {
        "team": "Liverpool",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 33,
        "minutes": 2377,
        "goals": 5,
        "assists": 3
    },
    "Ousmane Dembele": {
        "team": "PSG",
        "league": "Ligue 1",
        "season": "2025-26",
        "matches": 22,
        "minutes": 1064,
        "goals": 10,
        "assists": 7
    },
    "Michael Olise": {
        "team": "Bayern Munich",
        "league": "Bundesliga",
        "season": "2025-26",
        "matches": 32,
        "minutes": 2316,
        "goals": 15,
        "assists": 19
    },
    "Bruno Fernandes": {
        "team": "Manchester United",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 35,
        "minutes": 3063,
        "goals": 9,
        "assists": 21
    },
    "Luis Diaz": {
        "team": "Bayern Munich",
        "league": "Bundesliga",
        "season": "2025-26",
        "matches": 32,
        "minutes": 2449,
        "goals": 15,
        "assists": 14
    },
    "Lautaro Martinez": {
        "team": "Inter Milan",
        "league": "Serie A",
        "season": "2025-26",
        "matches": 30,
        "minutes": 2158,
        "goals": 17,
        "assists": 6
    },
    "Joao Pedro": {
        "team": "Chelsea",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 35,
        "minutes": 2661,
        "goals": 15,
        "assists": 5
    },
    "Ollie Watkins": {
        "team": "Aston Villa",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 37,
        "minutes": 2838,
        "goals": 16,
        "assists": 3
    },
    "Federico Dimarco": {
        "team": "Inter Milan",
        "league": "Serie A",
        "season": "2025-26",
        "matches": 35,
        "minutes": 2698,
        "goals": 7,
        "assists": 16
    },
    "Jarrod Bowen": {
        "team": "West Ham",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 38,
        "minutes": 3403,
        "goals": 9,
        "assists": 11
    },
    "Morgan Gibbs-White": {
        "team": "Nottingham Forest",
        "league": "Premier League",
        "season": "2025-26",
        "matches": 37,
        "minutes": 3098,
        "goals": 15,
        "assists": 4
    }
}

def dict_contents():
  contents = ''
  for position, players in soccer_players.items():
    contents += f'\n{position}\n'
    for stat, value in players.items():
      contents += f'    {stat.title()}: {value}\n'
  return contents

# list all players
def listAll(dicti):
    return dicti

# add a new player
def addPlayer(name, team, league, season, matches, minutes, goals, assists):
    soccer_players[name] = {
        'team': team,
        'league': league,
        'season': season,
        'matches': matches,
        'minutes': minutes,
        'goals': goals,
        'assists': assists
    }

# calculate average minutes per match
def avgMin(dicti):
    minutes = 0
    matches = 0

    for info in dicti.values():
        minutes += info['minutes']
        matches += info['matches']

    if matches == 0:
        return 0

    return round(minutes / matches, 2)

# calculate average goals for each team it adds up
def avgGoals(dicti):
    teams = {}
    counts = {}

    for info in dicti.values():
        team = info['team']

        if team not in teams:
            teams[team] = 0 #adds the temes to the new dictinary if the teams are not there
            counts[team] = 0 #calulcates players for team

        teams[team] += info['goals'] #adds up all the goals
        counts[team] += 1 #adds the players

    # divide goals by number of players
    for team in teams:
        teams[team] = round(teams[team] / counts[team], 2) #using average forumla to caluclate avg goals for each team

    return teams

# rank players by goal contributions. it adds the goals and assits and finds their average goals + assits per 90 min, then ranks who has the highest score by that.
def gapr(dicti):
    rankings = []

    for name, info in dicti.items():
        total = info['goals'] + info['assists']

        # calculate contributions per 90 minutes
        if info['minutes'] > 0:
            per90 = round(total / info['minutes'] * 90, 2)
        else:
            per90 = 0

        rankings.append([name, total, per90])

    # sort from highest to lowest, the player[2] means the second value in this case it is total goal contributions
    rankings.sort(key=lambda player: player[2] , reverse=True)

    return rankings

# main menu
print('\nWelcome to the ultimate soccer database!')

#while true loop for infinite amount of options until they stop
while True:
    print('\n1) List Players')
    print('2) Add Player')
    print('3) Delete Player')
    print('4) Rank Players')
    print('5) Calculate Average Goals per Team')
    print('6) Calculate Average Minutes Played Per Match')
    print('7) Exit the program\n')

    while True:
        try:
            choice = int(input('Enter Your Option: '))
            break
        except ValueError:
            print('This is not a correct option, please try again! ')

    if choice == 1:
        print(dict_contents())


    # get information for a new player
    elif choice == 2:
        name = (input('Enter the name of the player: ')).title()
        team = (input('What team do they play for: ')).title()
        league = input('What league do they play for: ')
        season = input('What season did they play in: ')
        matches = int(input('How many matches: '))
        minutes = int(input('How many minutes: '))
        goals = int(input('How many goals: '))
        assists = int(input('How many assists: '))

        addPlayer(name, team, league, season, matches, minutes, goals, assists)
        print('Added!')

    # delete a player by name
    elif choice == 3:
        name = (str(input('What player would you like to delete: '))).lower().title()
        if name in soccer_players:
            soccer_players.pop(name)
            print('Player deleted!')
        else:
            print('Player not found!')

    # display player rankings
    elif choice == 4:
        print('\nPlayer Rankings:')
        rank = 1

        for player in gapr(soccer_players):
            print(rank, '|', player[0], '| G+A:', player[1], '| Per 90:', player[2])
            rank += 1

    # display average goals for each team
    elif choice == 5:
        print('\nAverage Goals per Player for Each Team:')

        for team, average in avgGoals(soccer_players).items():
            print(team, ':', average)

    # display average minutes per match
    elif choice == 6:
        print('Average Minutes Per Match:', avgMin(soccer_players))

    # exit the program
    elif choice == 7:
        print('Goodbye!')
        break

    else:
        print('Invalid option!')
```
